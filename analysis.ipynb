**Analyse der erstellten Daten:**

```python
import json
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

sns.set(style="whitegrid")
```

```python
RESULT_FILE = "results/results_direct_gpt-5.1.json"

with open(RESULT_FILE, "r") as f:
    data = json.load(f)

df = pd.DataFrame(data)

print("Number of tasks:", len(df))
df.head()
```

```python
accuracy = df["correct"].mean()

print("Overall Accuracy:", round(accuracy * 100, 2), "%")
```

```python
strategy_acc = df.groupby("strategy")["correct"].mean().reset_index()

strategy_acc
```
